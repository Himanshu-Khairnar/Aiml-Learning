```python
product = ["a","b","c","d"]
sales = [10,15,7,12]
plt.bar(product,sales,color='black',label='2025 sales data')
plt.title("Product sales")
plt.xlabel("Product")
plt.ylabel("Sales per product")
plt.legend()
plt.show()
```

```python
product = ["a","b","c","d"]
sales = [10,15,7,12]
plt.pie(sales,labels=product,colors=['gray','orange','blue','green'],autopct='%1.1f%%',startangle=140)
plt.title("Product sales")

plt.legend()
plt.show()
```

```python
fig,ax = plt.subplots(1,2,figsize=(10,5))
x = [1,2,3,4]
y = [10,13,15,60]
ax[0].plot(x,y,color='blue')
ax[0].set_title("line plot")
ax[0].set_xlabel("X-axis Label")
ax[0].set_ylabel("Y-axis Label")
ax[1].set_title("bar plot")
ax[1].bar(x,y,color='red')
ax[1].set_xlabel("X-axis Label")
ax[1].set_ylabel("Y-axis Label")
fig.suptitle("comparison of Line and bar charts")

plt.tight_layout()

plt.savefig('saveFile.png',dpi =300,bbox_inches='tight')
```
