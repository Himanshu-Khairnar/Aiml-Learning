```python
import numpy as np
import pandas as pd
import seaborn as sns
np.random.seed(10)

patients = 60

patient_id = np.arange(1, patients + 1)
height = np.random.randint(150, 190, patients)
weight = np.random.randint(45, 110, patients)
systolic_bp = np.random.randint(100, 180, patients)
diastolic_bp = np.random.randint(60, 110, patients)
age = np.random.randint(20, 101, patients)

height_m = height / 100
bmi = weight / (height_m ** 2)

data = pd.DataFrame({
    "patient_id": patient_id,
    "height_cm": height,
    "weight_kg": weight,
    "systolic_bp": systolic_bp,
    "diastolic_bp": diastolic_bp,
    "bmi": bmi,
    "age": age
})

```

```python

data["high_risk"] = (data["bmi"] >= 30) | (data["systolic_bp"] >= 140) | (data["diastolic_bp"] >= 90)

high_risk_patients = data[data["high_risk"] == True]
sns.histplot(high_risk_patients["bmi"], bins=10, kde=True)
```
