```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import matplotlib.image as mpimg
```

```python
img_array = mpimg.imread('Askelad.webp')
print(img_array.shape)
print(img_array.dtype)
print(img_array)
plt.imshow(img_array)
```

```python
red_channel = img_array[:,:,0]
green_channel = img_array[:,:,1]
blue_channel = img_array[:,:,2]

fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))

ax1.imshow(red_chanel,cmap='Reds')
ax2.imshow(green_channel,cmap='Greens')
ax3.imshow(blue_chanel,cmap='Blues')
```

```python
#$$Gray = 0.299 \cdot R + 0.587 \cdot G + 0.114 \cdot B$$
gray = (0.299*red_channel) + (0.587*green_channel) + (0.114 *blue_channel)
plt.imshow(gray)
plt.show()
```

```python
(heigth, width, size) = img_array.shape
centery = heigth // 2
centerx = width // 2
newImag = img_array[centery - 100 : centery + 100, centerx - 100 : centerx + 100, :]
plt.imshow(newImag)
plt.show()
```

```python
negativeImage = 255-img_array
plt.imshow(negativeImage)
plt.show()
```

```python
binary  = gray>128
plt.imshow(binary)
```

```python
images = [img_array, red_channel, green_channel, blue_channel, gray, negativeImage, binary]
titles = ['Original', 'Red', 'Green', 'Blue', 'Gray', 'Negative', 'Binary']
cmaps = [None, 'Reds', 'Greens', 'Blues', 'gray', None, 'gray']

fig, axes = plt.subplots(1, 7, figsize=(21, 3.5))


for ax,img,t,c in zip(axes,images,titles,cmaps):
    ax.imshow(img,cmap=c)
    ax.set_title(t)
plt.tight_layout()

plt.savefig("images_converstion",dpi=300)
```
