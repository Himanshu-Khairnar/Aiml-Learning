```python
import numpy as np

np.random.seed(42)

n_samples = 200

study_hours = np.random.uniform(0, 10, n_samples)
attendance = np.random.uniform(50, 100, n_samples)
previous_marks = np.random.uniform(30, 100, n_samples)
sleep_hours = np.random.uniform(4, 9, n_samples)

score = (
    0.4 * study_hours +
    0.3 * (attendance / 10) +
    0.2 * (previous_marks / 10) +
    0.1 * sleep_hours
)

probability = 1 / (1 + np.exp(-score))
y = (probability > 0.5).astype(int)

X = np.column_stack((study_hours, attendance, previous_marks, sleep_hours))

X = (X - X.mean(axis=0)) / X.std(axis=0)

m, n = X.shape

w = np.zeros(n)
b = 0

learning_rate = 0.1
epochs = 1000

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def compute_loss(y, y_pred):
    return -np.mean(y * np.log(y_pred + 1e-9) + (1 - y) * np.log(1 - y_pred + 1e-9))

for i in range(epochs):
    z = np.dot(X, w) + b
    y_pred = sigmoid(z)

    dw = (1/m) * np.dot(X.T, (y_pred - y))
    db = (1/m) * np.sum(y_pred - y)

    w -= learning_rate * dw
    b -= learning_rate * db

    if i % 100 == 0:
        loss = compute_loss(y, y_pred)
        print("Epoch:", i, "Loss:", loss)

z = np.dot(X, w) + b
y_pred = sigmoid(z)
predictions = (y_pred >= 0.5).astype(int)

accuracy = np.mean(predictions == y)
print("\nFinal Accuracy:", accuracy)
```
