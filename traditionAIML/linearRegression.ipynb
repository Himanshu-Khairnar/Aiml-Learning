```python
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 4, 5])

w = 0.0
b = 0.0
lr = 0.01
epochs = 1000
n = len(x)

for _ in range(epochs):
    y_pred = w * x + b
    dw = (-2/n) * np.sum(x * (y - y_pred))
    db = (-2/n) * np.sum(y - y_pred)
    w = w - lr * dw
    b = b - lr * db

print(w, b)

x_new = 6
y_new = w * x_new + b
print(y_new)
```
