```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
area = np.array([500,800,1000,1200,1500,1800,2000])
price = np.array([25,40,50,60,75,90,100])


area_mean = np.mean(area)
area_std = np.std(area)
area_scaled = (area - area_mean) / area_std

w=0
b=0
lr=0.1
epochs=1000
n=len(area)

for i in range(epochs):
    y_pre = w*area_scaled + b
    dw = (1/n) * np.dot(area_scaled.T, (y_pre - price))
    db = (1/n) * np.sum(y_pre - price)
    w = w - lr*dw
    b = b - lr*db
print(f"Weight: {w}, Bias: {b}")

test_area = 3000

test_area_scaled = (test_area - area_mean) / area_std
prediction = w*test_area_scaled + b
print(f"Predicted price for {test_area} sq ft: {prediction}")

plt.scatter(area, price, color='blue', label='Actual Price')
plt.plot(area, w*area_scaled + b, color='red', label='Predicted Price')
plt.xlabel('Area (sq ft)')
plt.ylabel('Price (in $1000s)')
plt.title('House Price Prediction')
plt.legend()
plt.show()

```
