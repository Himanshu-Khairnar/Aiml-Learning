```python
import seaborn as sns
```

```python
tip = sns.load_dataset('tips')
sns.scatterplot(x='total_bill', y='tip', data=tip, hue='day', size='size',palette='YlGnBu')
```

```python
sns.histplot(tip['total_bill'], kde=True, bins=30)
# sns.displot(tip['total_bill'], kde=True, bins=30) deprecated
```

```python
sns.barplot(x='sex', y='tip', data=tip,palette='YlGn' )
```

```python
sns.boxplot(x='day', y='total_bill', data=tip ,hue='sex',palette='YlGnBu')
```

```python
sns.stripplot(
    x="day", y="total_bill", data=tip, jitter=True, hue="sex", palette="YlGnBu",dodge=True
)
```

```python
sns.jointplot(x='total_bill', y='tip', data=tip, kind='kde',fill=True, color='blue')
```

```python
sns.pairplot(tip, hue='day', palette='YlGnBu')
```

```python
titanic = sns.load_dataset('titanic')
sns.heatmap(titanic.corr(numeric_only=True), annot=True, cmap='YlGnBu')   
```

```python
import scipy

sns.clustermap(titanic.corr(numeric_only=True).fillna(0), annot=True, cmap='YlGnBu')
```
